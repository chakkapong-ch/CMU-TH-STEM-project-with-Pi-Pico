## Part 4 — Set Up the Raspberry Pi Pico W · 20 minutes
**Do this with the physical board and Thonny, outside Colab.**

### A. Install MicroPython firmware
Firmware is software stored on the board.

1. Obtain a stable **Pico W MicroPython UF2** file from your teacher or the [official Pico W download page](https://micropython.org/download/RPI_PICO_W/). Select **Pico W**, not Pico or Pico 2 W.
2. Disconnect USB and any other power source.
3. Press and hold the **BOOTSEL** button on the Pico W.
4. While holding it, connect the Pico W to the computer using a USB data cable.
5. Release BOOTSEL. A drive named **RPI-RP2** should appear.
6. Copy the `.uf2` file to this drive and wait for the transfer to finish.
7. The board restarts and the drive normally disappears.

### B. Connect in Thonny
1. Open Thonny.
2. Select the MicroPython interpreter for Raspberry Pi Pico/Pico W using the interpreter selector or interpreter settings.
3. Select the board's port if necessary.
4. Click **Stop/Restart** if needed.
5. Look for the MicroPython startup information and `>>>` prompt.

The exact interpreter label depends on the Thonny version. Ask your teacher to confirm that you are connected to the board.

## Quick troubleshooting reference
| Problem | What to check |
|---|---|
| Installer cannot proceed | Computer compatibility and installation permissions; ask your teacher. |
| RPI-RP2 does not appear | USB data cable, USB port, and BOOTSEL sequence. |
| Drive disappears after firmware transfer | This is normally expected when the board restarts. |
| Thonny uses desktop Python | Select the Pico MicroPython interpreter and board port. |
| Board does not respond | Check connections; close other applications using its port; try Stop/Restart. |
| SyntaxError | Check quotation marks, parentheses, and the highlighted line. |
| NameError | Check spelling and whether a variable was defined before use. |
| Colab runs code but Pico W does nothing | Colab's cloud runtime does not run the program on your board. Use Thonny. |
